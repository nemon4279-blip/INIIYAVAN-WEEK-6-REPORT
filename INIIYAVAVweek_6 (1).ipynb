```python
df = pd.read_csv("/content/healthcare_dataset.csv")
```

```python
df["Date of Admission"]=pd.to_datetime(df["Date of Admission"])
```

```python
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df["Admission Type"] = df["Admission Type"].str.strip()
```

```python
#Gender distribution

sns.countplot(data=df, x="Gender")
plt.title("Patient Distribution by Gender")

plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
age_bins = [-1, 17, 39, 59, float("inf")]

age_labels = [
    "Child",
    "Young Adult",
    "Middle Adult",
    "Senior"
]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels
)
```

```python
df["Age Group"].value_counts()
```

```python
df["Age Group"].value_counts()
```

```python
import matplotlib.pyplot as plt

sns.countplot(data=df, y="Medical Condition")
plt.title("Patient Distribution by Medical Condition")

plt.xlabel("Number of Patients")
plt.ylabel("Medical Condition")
plt.xticks(rotation=45)

plt.show()
```

```python
#medical condition vs gender
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=df, x="Medical Condition", hue="Gender")

plt.title("Medical Condition Distribution by Gender")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")

plt.show()
```

```python
#Billing distribution

sns.histplot(
    data=df,
    x="Billing Amount",
    bins=20,
    kde=True
)

plt.title("Distribution of Billing Amount")
plt.xlabel("Billing Amount")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Billing by Admission type
df["Admission Type"].value_counts()
```

```python
sns.boxplot(data=df, x="Admission Type", y="Billing Amount")
plt.title("Billing Amount Distribution by Admission Type")
plt.show()
```

```python
sns.violinplot(data=df, x="Medical Condition", y="Billing Amount")

plt.title("Billing Amount by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")

plt.xticks(rotation=45)

plt.show()
```

```python
# Average billing amount by medical condition
df.groupby("Medical Condition")["Billing Amount"].mean()
```

```python
df["Admission Year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month
```

```python
#Count Monthly Admissions

monthly_admissions = (
    df.groupby("Admission Month")
      .size()
      .reset_index(name="Admissions")
)
```

```python
#Admission trend

sns.lineplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Admissions",
    marker="o"
)

plt.title("Monthly Admissions Trend")
plt.xlabel("Month")
plt.ylabel("No of Admissions")

plt.xticks(rotation=45)

plt.show()
```

```python
#compare admission type over time

# Group by Month and Admission Type, then count
monthly_admissions_type = (
    df.groupby(["Admission Month", "Admission Type"])
      .size()
      .reset_index(name="Admissions")
)

```

```python
monthly_admissions_type.head()
```

```python
sns.lineplot(
    data=monthly_admissions_type,
    x="Admission Month",
    y="Admissions",
    hue="Admission Type",
    marker="o"
)

plt.title("Monthly Admissions by Admission Type")
plt.show()
```

```python
# Calculate Length of Stay in days
df["Length of Stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days

# Create the scatter plot
sns.scatterplot(
    data=df,
    x="Length of Stay",
    y="Billing Amount"
)

plt.title("Length of Stay vs Billing Amount")
plt.xlabel("Length of Stay (Days)")
plt.ylabel("Billing Amount")
plt.show()
```

```python
numeric_cols = [
    "Age",
    "Length of Stay",
    "Billing Amount"
]

corr = df[numeric_cols].corr()
corr
```
